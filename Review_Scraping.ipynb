## Wysa: Anxiety, therapy chatbot

Link: https://play.google.com/store/apps/details?id=bot.touchkin

ID: bot.touchkin

```python
id1 = 'bot.touchkin'
```

```python
wysa_details = app(id1, lang='en', country='us')
#wysa_details
```

```python
# Count/No. of reviews
wysa_details['reviews']
```

```python
wysa_reviews = reviews_all(
    id1,
    # sleep_milliseconds = 0, # defaults to 0
    # lang = 'en', # defaults to 'en'
    # country = 'us', # defaults to 'us'
    # sort = Sort.MOST_RELEVANT, # defaults to Sort.MOST_RELEVANT
    # count = 3, # defaults to 100
    # filter_score_with = 5, # defaults to None(means all score)
)
#wysa_reviews
```

```python
wysa_all = pd.json_normalize(wysa_reviews)
wysa_all.head()
```

```python
wysa_all.isnull().sum()
```

```python
wysa_all['content'].count()
```

```python
wysa_all.to_csv('review_data/wysa_reviews.csv', index=False)
```

```python
all_reviews = wysa_all[['content', 'score']]
all_reviews['source'] = 'wysa'
all_reviews.head(10)
```

```python
all_reviews.columns
```

```python
# renaming
all_reviews.rename(columns={'content': 'review', 'score': 'sentiment'}, inplace=True)
all_reviews.head(10)
```

```python
print(len(all_reviews))
```

```python
all_reviews.to_csv('all_reviews.csv', index=False)
```

## Conversion

```python
# We apply this formula after collecting all reviews

def map_sentiment(value):
    if value in [4, 5]:
        return 2
    elif value == 3:
        return 1
    elif value in [1, 2]:
        return 0
```

```python
all_reviews['sentiment'] = all_reviews['sentiment'].apply(map_sentiment)
all_reviews.tail(15)
```

```python
all_reviews.info()
```
